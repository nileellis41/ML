# 06 — Final Comparison

Master comparison table and visualisations across all models and variants.

```python
import sys; sys.path.insert(0, '..')
import pandas as pd, numpy as np, matplotlib.pyplot as plt
from src.utils.seeds import set_all_seeds
set_all_seeds()
%matplotlib inline
```

## 1. Generate Master Comparison Table

```python
from src.evaluation.reporting import run_all
table = run_all(rf=0.0)
table
```

## 2. Summary Statistics

```python
if not table.empty:
    key_cols = ['ann_return', 'sharpe', 'max_drawdown', 'ann_vol', 'turnover']
    avail = [c for c in key_cols if c in table.columns]
    display_table = table[avail].sort_values('sharpe', ascending=False)
    print(display_table.to_string(float_format='{:.4f}'.format))
```

## 3. Equity Curves — Base vs Regime Comparison

```python
from PIL import Image
from pathlib import Path

for fname in ['equity_curves_base.png', 'equity_curves_regime.png', 'drawdowns.png']:
    p = Path(f'../results/figures/{fname}')
    if p.exists():
        img = Image.open(p)
        plt.figure(figsize=(16, 5))
        plt.imshow(img)
        plt.axis('off')
        plt.title(fname)
        plt.show()
    else:
        print(f'{fname} not found -- run reporting first.')
```

## 4. Regime-Conditional Sharpe: Base vs Regime Variants

```python
if not table.empty:
    regime_cols = [c for c in table.columns if 'sharpe' in c]
    if regime_cols:
        fig, ax = plt.subplots(figsize=(12, 5))
        table[regime_cols].plot.bar(ax=ax, alpha=0.8)
        ax.set_title('Sharpe Ratios by Model, Variant, and Regime')
        ax.set_ylabel('Sharpe Ratio')
        ax.axhline(0, color='black', linewidth=0.5)
        ax.legend(fontsize=8)
        plt.tight_layout()
        plt.show()
```

## 5. Regime-Shaded SPY

```python
p = Path('../results/figures/regime_shaded_spy.png')
if p.exists():
    img = Image.open(p)
    plt.figure(figsize=(16, 5))
    plt.imshow(img)
    plt.axis('off')
    plt.show()
```

## 6. Export Master Table to Markdown

```python
md_path = Path('../results/tables/master_comparison.md')
if md_path.exists():
    with open(md_path) as f:
        print(f.read())
else:
    print('Markdown table not found -- run run_all() above first.')
```
